## food group glycan heatmap

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.gridspec import GridSpec

# Create a mapping dictionary for the renaming
rename_dict = {
    'Free_Kestose': 'Kestose',
    'Free_Raffinose': 'Raffinose',
    'Free_Maltotetraose': 'Maltotetraose',
    'Free_Stachyose': 'Stachyose',
    'Free_Maltopentaose': 'Maltopentaose',
    'Free_Verbascose': 'Verbascose',
    'Free_Maltohexaose': 'Maltohexaose',
    'Mono_Galactose': 'Galactose',
    'Mono_Fructose': 'Fructose',
    'Mono_Xylose': 'Xylose',
    'Mono_Arabinose': 'Arabinose',
    'Mono_Fucose': 'Fucose',
    'Mono_Rhamnose': 'Rhamnose',
    'Mono_GalA': 'GalA',
    'Mono_Mannose': 'Mannose',
    'Mono_Ribose': 'Ribose'
}

# Rename the columns
glycan_diet = glycan_diet.rename(columns=rename_dict)

# Define glycan types
polysaccharides = ['Arabinan', 'Galactan', 'Xyloglucan', 'Xylan', 'Mannan',
                   'Beta-Glucan', 'Chitin', 'Cellulose', 'Arabinoxylan',
                   'Galactomannan']

oligosaccharides = ['Kestose', 'Raffinose', 'Maltotetraose', 'Stachyose',
                    'Maltopentaose', 'Verbascose', 'Maltohexaose']

monosaccharides = ['Galactose', 'Fructose', 'Xylose', 'Arabinose', 'Fucose', 
                   'Rhamnose', 'GalA', 'Mannose', 'Ribose']

digesitble_sugars = ['Free_Lactose(trehalose)', 'Free_Maltose', 'Starch', 'Free_Fructose', 'Free_Glucose', 'Free_Sucrose', 'Mono_Glucose']

all_glycans = polysaccharides + oligosaccharides + monosaccharides

# Step 1: Sum glycan intake per person per food
person_food_glycans = glycan_diet.groupby(['SEQN', 'simple_name', 'food_group'])[all_glycans].sum().reset_index()

# Step 2: Average across all people who consumed each food
glycan_food_matrix = person_food_glycans.groupby(['simple_name', 'food_group'])[all_glycans].mean().reset_index()

# Step 3: Aggregate to food group level
glycan_foodgroup_matrix = glycan_food_matrix.groupby('food_group')[all_glycans].sum().reset_index()

# Sort by total glycan content
glycan_foodgroup_matrix['total'] = glycan_foodgroup_matrix[all_glycans].sum(axis=1)
glycan_foodgroup_matrix = glycan_foodgroup_matrix.sort_values('total', ascending=False).drop('total', axis=1)

# Prepare data
food_groups = glycan_foodgroup_matrix['food_group'].values

# Calculate width ratios based on number of glycans in each category
width_ratios = [len(polysaccharides), len(oligosaccharides), len(monosaccharides)]

# Create figure with proportional widths
fig = plt.figure(figsize=(6, 4))
gs = GridSpec(1, 3, width_ratios=width_ratios, wspace=0.3)

axes = [fig.add_subplot(gs[0]), fig.add_subplot(gs[1]), fig.add_subplot(gs[2])]

glycan_types = [
    (polysaccharides, 'Polysaccharides', axes[0]),
    (oligosaccharides, 'Oligosaccharides', axes[1]),
    (monosaccharides, 'Monosaccharides', axes[2])
]

for i, (glycan_list, title, ax) in enumerate(glycan_types):
    data = glycan_foodgroup_matrix[glycan_list].values
    
    im = sns.heatmap(data,
                xticklabels=glycan_list,
                yticklabels=food_groups if i == 0 else False,  # Only first plot
                cmap='YlOrRd',
                cbar_kws={
                    'label': 'Average Intake (g per day)', 
                    'shrink': 0.3,
                    'aspect': 8,
                    'pad': 0.04
                },
                ax=ax,
                linewidths=0.5,
                linecolor='white',
                square=True)
    
    # Scale down colorbar text
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=4)
    cbar.set_label('Average Intake (g per day)', fontsize=5)
    
    ax.set_title(title, fontsize=8, pad=5)
    ax.set_xlabel('', fontsize=10)
    ax.tick_params(axis='x', rotation=90, labelsize=5)
    
    # Only show y-label and y-ticks on first plot
    if i == 0:
        ax.set_ylabel('Food Group', fontsize=10)
        ax.tick_params(axis='y', labelsize=6, rotation=0)
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', left=False)  # Remove ticks

plt.suptitle('', 
             fontsize=9, fontweight='bold', y=0.98)

plt.tight_layout()
plt.savefig('/Users/jules.larke/work/project/pulse_update_copy/plot/01/glycan_profile/diet_intake/glycan_foodgroup_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Get the sum of each glycan column across all food groups
glycan_totals = glycan_foodgroup_matrix.drop('food_group', axis=1).sum()

# Get the values for "Beans, Peas, Lentils (Pulses)"
pulses_row = glycan_foodgroup_matrix[glycan_foodgroup_matrix['food_group'] == 'Beans, Peas, Lentils (Pulses)'].drop('food_group', axis=1)

# Calculate the percentage contribution for each glycan
pulses_percentage = (pulses_row.values[0] / glycan_totals.values) * 100

# Create a dataframe to view the results
import pandas as pd
results = pd.DataFrame({
    'Glycan': glycan_totals.index,
    'Total': glycan_totals.values,
    'Pulses_Amount': pulses_row.values[0],
    'Pulses_Percentage': pulses_percentage
})

print(# designating glycan sources that are >20% to be pulse enriched sources
results.sort_values(by='Pulses_Percentage', ascending=False))
```

## plot food composition (DFG2 data)

```python


# --- 1. Define Polysaccharides ---
polysaccharides = [
    'Arabinan', 'Galactan', 'Xyloglucan', 'Xylan', 'Mannan', 'Beta-Glucan',
    'Chitin', 'Cellulose', 'Arabinoxylan', 'Galactomannan'
]

def plot_food_polysaccharides_stacked(df):
    """
    Aggregates data per participant, then plots a single stacked bar chart
    comparing polysaccharide composition across food groups.
    """
    # Define aesthetics
    LINE_WIDTH = 1.2
    BAR_WIDTH = 0.7
    
    # --- 2. Data Aggregation ---
    # First: Sum all polysaccharides per food_group
    df_food_group = df.groupby(['food_group'])[polysaccharides].mean().reset_index()
    
    # Calculate total per food_group for the SEM calculation
    df_food_group['Total_Poly'] = df_food_group[polysaccharides].sum(axis=1)

    # Second: Calculate the MEAN across all participants for each food group
    df_plot = df_food_group.groupby('food_group')[polysaccharides].mean()
    
    # Calculate total mean intake for each food group to determine sorting order
    total_means = df_plot.sum(axis=1).sort_values(ascending=False)
    
    # Reorder the plotting dataframe based on the sorted total means
    df_plot = df_plot.loc[total_means.index]
    
    # Calculate the SEM of the total intake, ordered the same way
    sem_values = df_food_group.groupby('food_group')['Total_Poly'].sem().loc[total_means.index]
    
    # --- 3. Visualization ---
    fig, ax = plt.subplots(figsize=(6, 6))
    x_pos = np.arange(len(df_plot.index))
    
    # 1. Error Bars (Plotted behind, top-half only)
    # yerr=[zeros, SEM] ensures the bar only extends upwards
    ax.errorbar(
        x_pos, 
        total_means.values, 
        yerr=[[0]*len(sem_values), sem_values.values], 
        fmt='none', 
        color='black', 
        capsize=6, 
        linewidth=LINE_WIDTH, 
        zorder=1
    )

    # 2. Black border for the FULL bar (Plotted behind the stacks)
    ax.bar(
        x_pos, 
        total_means.values, 
        width=BAR_WIDTH, 
        color='none', 
        edgecolor='black', 
        linewidth=LINE_WIDTH, 
        zorder=1.5
    )

    # 3. Stacked Bar Chart
    df_plot.plot(
        kind='bar', 
        stacked=True, 
        ax=ax, 
        width=BAR_WIDTH,
        edgecolor=None, 
        linewidth=0, 
        zorder=2,
        legend=False # Custom legend handling below
    )

    # --- 4. Formatting ---
    ax.set_title('', fontsize=12, pad=20)
    ax.set_ylabel('Average Polysaccharide content (g per 100g)', fontsize=12)
    ax.set_xlabel('Food Group', fontsize=12)
    ax.set_xticks(x_pos)
    ax.set_xticklabels(df_plot.index, rotation=45, ha='right')

    # --- 5. Legend Handling (Removing SEM) ---
    handles, labels = ax.get_legend_handles_labels()
    # Ensure only polysaccharide components are in the legend
    filtered_handles = [h for h, l in zip(handles, labels) if l in polysaccharides]
    filtered_labels = [l for l in labels if l in polysaccharides]

    ax.legend(
        filtered_handles, 
        filtered_labels, 
        title='Polysaccharides', 
        bbox_to_anchor=(1.05, 1), 
        loc='upper left',
        frameon=False
    )

    plt.tight_layout()
    plt.savefig('../../plot/01/glycan_profile/food_composition/dfg_polysaccharide_food_comp.png', dpi=300)
    plt.show()
```

```python
plot_food_polysaccharides_stacked(dfg2_unique)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Define Polysaccharides ---
polysaccharides = [
    'Arabinan', 'Galactan', 'Xyloglucan', 'Xylan', 'Mannan', 'Beta-Glucan',
    'Chitin', 'Cellulose', 'Arabinoxylan', 'Galactomannan'
]

def plot_food_polysaccharides_ordered(df):
    """
    Aggregates data per participant, sorts food groups by descending total intake,
    and plots a stacked bar chart with SEM error bars and full-bar borders.
    """
    # Define aesthetic constants
    LINE_WIDTH = 1.2
    BAR_WIDTH = 0.7
    
    # --- 2. Data Aggregation ---
    # Sum all polysaccharides per participant within each food group
    df_participant = df.groupby(['SEQN', 'food_group'])[polysaccharides].sum().reset_index()
    
    # Calculate total per participant for the SEM calculation later
    df_participant['Total_Poly'] = df_participant[polysaccharides].sum(axis=1)

    # Calculate the MEAN intake for each glycan across participants
    df_plot = df_participant.groupby('food_group')[polysaccharides].mean()
    
    # Calculate total mean intake for each food group to determine sorting order
    total_means = df_plot.sum(axis=1).sort_values(ascending=False)
    
    # Reorder the plotting dataframe based on the sorted total means
    df_plot = df_plot.loc[total_means.index]
    
    # Calculate the SEM of the total intake, ordered the same way
    sem_values = df_participant.groupby('food_group')['Total_Poly'].sem().loc[total_means.index]
    
    # --- 3. Visualization ---
    fig, ax = plt.subplots(figsize=(12, 8))
    x_pos = np.arange(len(df_plot.index))
    
    # 1. Error Bars (Plotted behind, top-half only)
    ax.errorbar(
        x_pos, 
        total_means.values, 
        yerr=[[0]*len(sem_values), sem_values.values], 
        fmt='none', 
        color='black', 
        capsize=6, 
        linewidth=LINE_WIDTH, 
        zorder=1
    )

    # 2. Black border for the FULL bar (Plotted behind the stacks)
    ax.bar(
        x_pos, 
        total_means.values, 
        width=BAR_WIDTH, 
        color='none', 
        edgecolor='black', 
        linewidth=LINE_WIDTH, 
        zorder=1.5
    )

    # 3. Stacked Bar Chart (Composition)
    df_plot.plot(
        kind='bar', 
        stacked=True, 
        ax=ax, 
        width=BAR_WIDTH,
        edgecolor=None, # Removes borders between individual stack segments
        linewidth=0, 
        zorder=2,
        legend=False 
    )

    # --- 4. Formatting ---
    ax.set_title('', fontsize=16, pad=20)
    ax.set_ylabel('Average Intake g / day • 1000 kcal', fontsize=12)
    ax.set_xlabel('Food Group', fontsize=12)
    ax.set_xticks(x_pos)
    ax.set_xticklabels(df_plot.index, rotation=45, ha='right')

    # --- 5. Legend (Filtered for polysaccharides only) ---
    handles, labels = ax.get_legend_handles_labels()
    # Filter to exclude the SEM and border bar from the legend
    filtered_handles = [h for h, l in zip(handles, labels) if l in polysaccharides]
    filtered_labels = [l for l in labels if l in polysaccharides]

    ax.legend(
        filtered_handles, 
        filtered_labels, 
        title='Polysaccharides', 
        bbox_to_anchor=(1.05, 1), 
        loc='upper left',
        frameon=False
    )

    plt.tight_layout()
    #plt.savefig('food_group_polysaccharides_ordered.png', dpi=300)
    plt.show()
```

```python
plot_food_polysaccharides_ordered(glycan_diet)
```

## glycan intakes for food group specific contributions (checking data)

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Get unique food groups
food_groups = glycan_food_matrix['food_group'].unique()

# Layout: 3 columns, N rows
n_cols = 3
n_rows = -(-len(food_groups) // n_cols)  # ceiling division

fig, axes = plt.subplots(n_rows, n_cols,
                          figsize=(n_cols * 5, n_rows * 4))
axes = axes.flatten()

for i, group in enumerate(sorted(food_groups)):
    ax = axes[i]

    # Filter and get top 10 by Cellulose intake
    subset = (
        glycan_food_matrix[glycan_food_matrix['food_group'] == group]
        .nlargest(10, 'Cellulose')
        .sort_values('Cellulose', ascending=True)  # ascending for horizontal bar
    )

    ax.barh(
        subset['simple_name'],
        subset['Cellulose'],
        color='#3266ad',
        edgecolor='none'
    )

    ax.set_title(group, fontsize=11, fontweight='bold', pad=8)
    ax.set_xlabel('Cellulose (g/day)', fontsize=9)
    ax.tick_params(labelsize=8)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# Hide any unused subplots
for j in range(len(food_groups), len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Top 10 food contributors to cellulose intake by food group',
             fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('cellulose_by_foodgroup.png', dpi=150, bbox_inches='tight')
plt.show()
```
