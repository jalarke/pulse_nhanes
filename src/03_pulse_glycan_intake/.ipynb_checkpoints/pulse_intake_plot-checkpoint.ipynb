```python
import numpy as np

# --- Step 1: Get all unique pulse foods ---
pulse_foods = pulse_consumer['ingred_desc'].unique()

# --- Step 2: For consumers, fill in 0 for pulse foods they didn't eat ---
all_consumer_combos = pulse_consumer[['subject_id']].drop_duplicates().assign(key=1).merge(
    pd.DataFrame({'ingred_desc': pulse_foods, 'key': 1}), on='key'
).drop(columns='key')

consumers_full = all_consumer_combos.merge(
    pulse_consumer[['subject_id', 'ingred_desc', 'cup_eq']],
    on=['subject_id', 'ingred_desc'],
    how='left'
).fillna({'cup_eq': 0.0})
consumers_full['consumer'] = 'yes'

# --- Step 3: Cross-join non-consumers with all pulse foods (cup_eq = 0) ---
non_consumers_expanded = pulse_nonconsumer[['subject_id']].assign(key=1).merge(
    pd.DataFrame({'ingred_desc': pulse_foods, 'key': 1}), on='key'
).drop(columns='key')
non_consumers_expanded['cup_eq'] = 0.0
non_consumers_expanded['consumer'] = 'no'

# --- Step 4: Combine ---
combined = pd.concat([consumers_full, non_consumers_expanded], ignore_index=True)

# --- Step 5: Overall averages + SEM (consumers + non-consumers) ---
overall_stats = (combined.groupby('ingred_desc')['cup_eq']
                         .agg(
                             avg_cup_eq='mean',
                             sem_cup_eq=lambda x: x.std() / np.sqrt(len(x))
                         )
                         .reset_index())

# --- Step 6: Consumer-only averages + SEM (only rows where cup_eq > 0) ---
consumer_stats = (consumers_full[consumers_full['cup_eq'] > 0]
                         .groupby('ingred_desc')['cup_eq']
                         .agg(
                             avg_cup_eq_consumers='mean',
                             sem_cup_eq_consumers=lambda x: x.std() / np.sqrt(len(x))
                         )
                         .reset_index())

# --- Step 7: Merge ---
avg_intake = overall_stats.merge(consumer_stats, on='ingred_desc')
```

```python
avg_intake
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# --- Sort by consumer average intake descending ---
ordered_foods = avg_intake.sort_values('avg_cup_eq_consumers', ascending=False)['ingred_desc'].tolist()

# --- Reshape into long format ---
plot_df = pd.melt(
    avg_intake,
    id_vars='ingred_desc',
    value_vars=['avg_cup_eq', 'avg_cup_eq_consumers'],
    var_name='group',
    value_name='mean'
)

sem_df = pd.melt(
    avg_intake,
    id_vars='ingred_desc',
    value_vars=['sem_cup_eq', 'sem_cup_eq_consumers'],
    var_name='group',
    value_name='sem'
)
sem_df['group'] = sem_df['group'].replace({
    'sem_cup_eq': 'avg_cup_eq',
    'sem_cup_eq_consumers': 'avg_cup_eq_consumers'
})

plot_df = plot_df.merge(sem_df, on=['ingred_desc', 'group'])
plot_df['group'] = plot_df['group'].replace({
    'avg_cup_eq': 'All People',
    'avg_cup_eq_consumers': 'Consumers Only'
})

# --- Apply ordering ---
plot_df['ingred_desc'] = pd.Categorical(plot_df['ingred_desc'], categories=ordered_foods, ordered=True)
plot_df = plot_df.sort_values(['ingred_desc', 'group']).reset_index(drop=True)

# --- Plot ---
fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=plot_df,
    x='ingred_desc',
    y='mean',
    hue='group',
    hue_order=['All People', 'Consumers Only'],
    order=ordered_foods,
    palette='muted',
    ax=ax
)

# --- Calculate x positions explicitly ---
n_foods = len(ordered_foods)
n_hues = 2
group_order = ['All People', 'Consumers Only']
bar_width = 0.4  # seaborn default for 2 hue groups
offsets = [-bar_width / 2, bar_width / 2]

for i, group in enumerate(group_order):
    for j, food in enumerate(ordered_foods):
        row = plot_df[(plot_df['ingred_desc'] == food) & (plot_df['group'] == group)]
        if row.empty:
            continue
        x_pos = j + offsets[i]
        ax.errorbar(
            x=x_pos,
            y=row['mean'].values[0],
            yerr=row['sem'].values[0],
            fmt='none',
            color='black',
            capsize=4,
            linewidth=1.2
        )

ax.set_xlabel('Pulse Food', fontsize=12)
ax.set_ylabel('Average Intake (cup eq)', fontsize=12)
ax.set_title('Average Pulse Intake by Food and Group', fontsize=14)
ax.legend(title='Group')
sns.despine()

plt.tight_layout()
plt.savefig('pulse_intake_barplot.png', dpi=300)
plt.show()
```
