```python
import pandas as pd
```

```python
#load data for the four DFG2 glycan analyses and map of NHANES food descriptions to DFG2 descriptions
poly = pd.read_csv('../../data/00/dfg2/polysaccharide_v6.csv')
free = pd.read_csv('../../data/00/dfg2/free_saccharide_v6.csv')
free.drop(columns={'sample_id','food_group'},inplace=True)
mono = pd.read_csv('../../data/00/dfg2/monosaccharide_v6.csv')
mono.drop(columns={'sample_id','food_group'},inplace=True)
wweia_map = pd.read_csv('../../data/00/ingredient_matching/nhanes_dfg2_map_final.csv') # manually curated list of food mapping from NHANES to DFG2
```

```python
# rename columns to distinguish the free and monosaccharides of glucose, fructose
free.set_index('simple_name', inplace=True)
free.columns = ['Free_' + str(col) for col in free.columns]
free.reset_index(inplace=True)

mono.set_index('simple_name', inplace=True)
mono.columns = ['Mono_' + str(col) for col in mono.columns]
mono.reset_index(inplace=True)
```

```python
# merge the different DFG2 platforms into one dataframe
poly_free = poly.merge(free, on='simple_name', how='left')
dfg_all = poly_free.merge(mono, on='simple_name', how='left')
dfg_all = dfg_all.drop(columns=['sample_id'])
```

```python
# merge nhanes descriptions with DFG2 database
nhanes_dfg2_matches = wweia_map.merge(dfg_all, on='simple_name', how='left')
```

```python
nhanes_dfg2_matches.to_csv('../../data/00/wweia_dataset/nhanes_dfg2_matched.csv',index=None)
```
