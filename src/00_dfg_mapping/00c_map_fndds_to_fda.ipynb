### Title: 00c_map_fndds_to_fda
### Purpose: Clean up mapping file of FNDDS to FDA. Removes non-matches and replace with corrected matches during manual curation
### Date: July 3, 2025
### Author: Jules Larke

```python
import pandas as pd
```

```python
fda = pd.read_csv('../../data/00/ingredient_matching/fda_fndds_map_complete.csv', usecols=['fda_desc', 'fndds_desc', 'Ingredient code', 'match', 'replace'])
fndds_ingred = pd.read_csv('../../data/00/ingredient_matching/fndds_all_ingredient_nutrient_values_2023.csv', usecols=['Ingredient code', 'Ingredient description'])
```

```python
replace = fda[fda['replace'].notna()]
replace = replace[['fda_desc', 'replace']]
replace.rename(columns={'replace':'Ingredient description'},inplace=True)
replace = replace.merge(fndds_ingred, on='Ingredient description')
```

```python
matched = fda[fda['match'].isna()]
matched = matched.copy()
matched.rename(columns={'fndds_desc':'Ingredient description'},inplace=True)
matched = matched[['fda_desc', 'Ingredient description', 'Ingredient code']]
```

```python
fda_fndds = pd.concat([matched, replace])
```

```python
fda_fndds.sort_values('Ingredient code').to_csv('../../data/00/ingredient_matching/fda_fndds_map_final.csv',index=None)
```
